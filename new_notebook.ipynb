```sage
from sage.all import *
load("estimator.py")
print("LWE Estimator Loaded Successfully!")
```

```sage
def analysis(n, q, m):
    print(f"--- Security Analysis (n={n}, q=2^{log(q,2)}) ---")
    alpha = float(8/q)
    # 1. בדיקת סוד אחיד (Standard Secret)
    print("\n[Case 1] Uniform Secret Distribution:")
    res_uniform = estimate_lwe(n, alpha, q, m=m)

    # 2. בדיקת סוד דליל (Sparse Secret - נפוץ מאוד ב-HSS)
    h = 80 # מספר איברים שאינם אפס
    print(f"\n[Case 2] Sparse Secret (Hamming Weight h={h}):")
    res_sparse = estimate_lwe(n, alpha, q, secret_distribution = ((-1,1), h), m=m, skip=("mitm", "arora-gb", "bkw", "dec"))
    print("\n\n")
```

```sage
analysis(10240, 2^276, 10240*276)
```

```sage
analysis(8192, 2^236, 8192*236)
```

```sage
analysis(7168, 2^276, 7168*276)
analysis(7680, 2^266, 7680*266)
analysis(6656, 2^226, 6656*226)
analysis(9728, 2^266, 9728*266)
analysis(8192, 2^236, 8192*236)
```
