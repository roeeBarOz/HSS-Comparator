```python
from sage.all import *
load("estimator.py")
print("LWE Estimator Loaded Successfully!")
```

```python
def estimate(n, alpha, q, h="None"):
    if h == "None":
        res = estimate_lwe(n, alpha, q)
    else:
        res = estimate_lwe(n, alpha, q, secret_distribution=(0, h))
```

```python
# הגדרות בסיסיות
n = 7000       # עבור RLWE זו הדרגה, עבור LWE זה הממד
q = 2^250       # מודולוס לדוגמה (32 ביט)
sd = 3.2       # סטיית תקן של הרעש
alpha = float(sd / q)

print(f"--- Security Analysis (n={n}, q=2^{log(q,2)}) ---")

# 1. בדיקת סוד אחיד (Standard Secret)
print("\n[Case 1] Uniform Secret Distribution:")
res_uniform = estimate_lwe(n, alpha, q)

# 2. בדיקת סוד דליל (Sparse Secret - נפוץ מאוד ב-HSS)
h = 64 # מספר איברים שאינם אפס
print(f"\n[Case 2] Sparse Secret (Hamming Weight h={h}):")
res_sparse = estimate_lwe(n, alpha, q, secret_distribution=(0, h))
```
